# External Validation
**Purpose:** Test whether the regimes match things we know independently: the 2020 lockdown, Diwali, stubble season, monsoon rain, weekdays, and a train/test split across years.

**Inputs:** `data/results/delhi_regimes.csv`, `delhi_context.csv`, `delhi_scaled.csv`

**Outputs:** none (findings written in the conclusion)

## 1. Load the data

```python
from pathlib import Path
import pandas as pd, numpy as np
from sklearn.cluster import KMeans
from sklearn.metrics import adjusted_rand_score
from scipy.stats import chi2_contingency

BASE = Path.cwd().parent
reg = pd.read_csv(BASE / "data" / "results" / "delhi_regimes.csv", parse_dates=["Date"], index_col="Date")
ctx = pd.read_csv(BASE / "data" / "features" / "delhi_context.csv", parse_dates=["Date"], index_col="Date")
df = reg.join(ctx[["is_weekend", "diwali_window", "stubble_season", "monsoon", "lockdown"]])
order = ["Monsoon washout", "Moderate winter smog", "Pre-monsoon dust", "Severe stagnant winter smog", "Warm, cleaner days"]
print(df.shape)
```

## 2. Lockdown: same calendar window (25 Mar to 31 May) in each year

```python
md = df.index.strftime("%m-%d")
win = (md >= "03-25") & (md <= "05-31")
print(pd.crosstab(df.loc[win, "year"], df.loc[win, "regime"], normalize="index").reindex(columns=order, fill_value=0).round(2))
print(df.loc[win].groupby("year")[["PM2.5", "PM10", "NO2", "CO"]].mean().round(1))
```

## 3. Diwali: PM2.5 on the day after versus the week before

```python
rows = []
for d in pd.to_datetime(["2017-10-19", "2018-11-07", "2019-10-27"]):
    pre = df["PM2.5"].loc[d - pd.Timedelta(days=7): d - pd.Timedelta(days=1)].mean()
    nxt = df["PM2.5"].get(d + pd.Timedelta(days=1), np.nan)
    rows.append({"date": d.date(), "pre_7d_mean": round(pre), "day_after": round(nxt),
                 "ratio": round(nxt / pre, 2), "regime_day_after": df["regime"].get(d + pd.Timedelta(days=1))})
print(pd.DataFrame(rows))
```

## 4. Stubble season (15 Oct to 30 Nov) versus Dec-Jan

```python
period = np.where(df["stubble_season"], "Oct15-Nov", np.where(df["month"].isin([12, 1]), "Dec-Jan", "other"))
print(df.groupby(period)[["PM2.5", "PM_ratio", "CO", "NO2", "wind_speed", "temp"]].mean().round(2))
print(pd.crosstab(period, df["regime"], normalize="index").reindex(columns=order, fill_value=0).round(2))
```

## 5. Do heavy-rain days land in the monsoon regime?

```python
heavy = df["rain"] > 10
print("days with rain > 10 mm:", int(heavy.sum()))
print(df.loc[heavy, "regime"].value_counts(normalize=True).round(2))
```

## 6. Weekday versus weekend

```python
tab = pd.crosstab(df["is_weekend"], df["regime"])
print((tab.div(tab.sum(axis=1), axis=0)).round(3))
print("chi-square p-value:", round(chi2_contingency(tab)[1], 3))
```

## 7. Stability across time: fit on 2016-2018, test on 2019-2020

```python
X = pd.read_csv(BASE / "data" / "features" / "delhi_scaled.csv", parse_dates=["Date"], index_col="Date")
Xn = X.drop(columns=["wind_dir_sin", "wind_dir_cos"])
km_all = pd.read_csv(BASE / "data" / "results" / "kmeans_labels.csv", parse_dates=["Date"], index_col="Date")["kmeans_cluster"]

train = Xn.index < "2019-01-01"
m = KMeans(n_clusters=5, n_init=20, random_state=42).fit(Xn[train])
pred = pd.Series(m.predict(Xn[~train]), index=Xn.index[~train])
print("train days:", int(train.sum()), "| test days:", int((~train).sum()))
print("ARI, train-only model vs full model on 2019-2020:", round(adjusted_rand_score(km_all[~train], pred), 3))
print(pd.crosstab(km_all[~train], pred))
```
